#### **Task_01**: A business center rents meeting rooms at the following rates: 
- a)	1 hour – 20
- b)	½ hour – 10
- c)	Unlimited hours in a day – 100
- Write a program to calculate the meeting room bill. The manager should enter the number of hours the room was used, or 0 for unlimited hours.

```python
#Task_01

hour_rate = 20 #cost for an hour
half_hour_rate = 10 ##cost for an 1/2 hour
unlimited_rate = 100 #cost for unlimited
whole_hours = 0 #initializing outside the if statement
bill = 0 #initializing outside the if statement
extra_time = 0 #initializing outside the if statement

hours_used = float(input("Enter the number of hours the room was used for (0 for unlimited) : "))

if hours_used == 0: #if unlimited time, charge 100
    bill = unlimited_rate
    print("Customer Charged for Unlimited Hours")
else:   #if not unlimited, calculate bill
    whole_hours = int(hours_used)   #how many Whole hours was the room used for
    extra_time = round(hours_used - whole_hours ,2)     #how much additional time was the room used for (e.g. 2 hours and 20 mins. 2 hours whole, 20 mins extra time)
    bill = whole_hours * hour_rate      #calculate the amount of cost for the whole hours

    if extra_time > 0.5:    #if extra time was more than 1/2 hour charge the whole hour rate
        bill = bill + hour_rate
    elif extra_time > 0:    #if any time less than 1/2 hour remains, charge the half hour rate
        bill = bill + half_hour_rate

    print(f"The Meeting Room was used for a total of {hours_used} hours")
    print(f"Whole Hours Used: {whole_hours}")
    print(f"Additional Hours Used: {extra_time}")

if bill > unlimited_rate:    #this is not stated so I am making an assumption that if the bill is greater than 100 we want to limit the bill
    bill = unlimited_rate
    print("Bill exceeded the unlimited rate, so unlimited rate was charged")

print(f"The total meeting room bill is: {bill}")
```

#### **Task_01:** Write a function Extract_Even(List) to return all even numbers from a list.
- List1 = [1, 2, 3, 4, 5, 6]
- print(Extract_Even(List1))
- [2, 4, 6]

```python
def Extract_Even(List):     #define the function
    even_numbers = []       #initialize an empty list
    for num in List:        #for every number in the list
        if num % 2 ==0:     #check if dividing by 2 leaves no remained (makign it even)
            even_numbers.append(num) #append to the list if true
    return even_numbers        #return the list

List1 = [1,2,3,4,5,6]           #define the list  
print(Extract_Even(List1))      #print the list the function returns
```

<h2 style="color: black;"><strong> Question 4: Reading Files and Exploring Data in Python.<span style="color: red;">[25 Points]</span></strong></h2>

#### The data shows the highest degrees earned by a sample of 200 chief executive officers (CEOs) in the United States.
- **Education Level:** Bachelor’s, Master’s, PhD, or None. “None” indicates that the CEO has not earned a college or university degree.
  - Read **CEO.csv** using the with statement. Complete **any one** of the following tasks:
      - **Task:** Use a **while loop** and **conditional statements** to count how many CEOs belong to each education level. Display the totals.
      - <h4 style="text-align: center; color: red;"><strong>OR</strong></h4>
      - **Task:** Use a **for loop** and **conditional statements** to count how many CEOs belong to each education level. Display the totals.

```python
#similar to the frost work

import csv  #import the csv module

degree_counts = {} #initialize an empty list 

with open("CEO.csv","r", encoding="utf-8") as file:
    reader = csv.reader(file) #Create a CSV reader
    headings = next(reader) #read headings
    responses = list(reader) #store student responses

    for row in responses: # for each response in the file
        degree = row[1] #look at response in second column

        if degree in degree_counts: #if the degree is already present, add 1 to the count
            degree_counts[degree] += 1
        else:
            degree_counts[degree] = 1 # if the degree had not been seen before start the count at 1

sorted_counts = sorted(degree_counts.items(), key=lambda item: item[1], reverse=True) #sort the pairs by count, highest first

print("CEO Degree Levels: ") #print the sorted counts
for degree, count in sorted_counts:
    print(f"{degree}: {count}")
            
```
