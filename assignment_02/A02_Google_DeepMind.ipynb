Task 01: Read the Dataset and Create a Database Connection

```python
import pandas as pd   ####Imports Pandas
import sqlite3 ###Import Sql
conn = sqlite3.connect("My_Data.db") # create/connect to the database
cursor = conn.cursor()

pd.read_csv('Course_Evaluation.csv').to_sql('evaluation',conn,if_exists = 'replace',index = False) # load CSV into SQLite as a table

pd.read_sql_query('''
select *
from evaluation
limit 5
''', conn) # select * returns all columns; limit 5 returns only the first five rows
```

### Business Interpretation
The dataset loaded successfully into the SQLite database as the evaluation table. One data quality issue stands out. One student’s AI hours were recorded as “2-Jan,” which likely happened when Excel converted a “1-2” range into a date. That value is stored as missing (NaN), so this student is left out of any calculation that averages hours.

Task 02: Which Seniors Use AI Most for Assignments?
</h3>

<p>
<strong>Question:</strong>
Which senior students reported using AI on assignments?
Display their <code>Major</code>, <code>AI_Tool</code>,
and <code>AI_Hours_Per_Week</code> for students who
answered <code>Yes</code> to <code>Used_AI_Assignment</code>.
Arrange the results so that the heaviest AI users appear first.

```python
pd.read_sql_query('''
select Major, AI_Tool, AI_Hours_Per_Week 
from evaluation
where Used_AI_Assignment = 'Yes'
    and Academic_Standing = 'Senior'
order by AI_Hours_Per_Week desc
''',conn)

# select pulls only the three columns the question asks for
# where keeps students who answered Yes to using AI AND are seniors
# order by desc sorts by weekly hours so the heaviest users appear first
```

### Business Interpretation
Among seniors who used AI on assignments, the heaviest user was a **Business Analytics** major using **Claude** for **10** hours per week. That is well above everyone else, who reported 4 hours or less. The 10-hour response is an outlier and could pull any group average upward. Seniors also use a wide mix of tools, and several reported using more than one.

Task 03: Which AI Tools Are Most Commonly Reported?

each tool entry, arranged from most to least common.

### Business Interpretation
**ChatGPT** is the most frequently reported entry, listed alone by 9 of the 20 students. Copilot and Claude follow with 2 each. Because many students listed more than one tool, the grouped counts understate ChatGPT's total use. ChatGPT appears in 15 of the 20 responses when combination entries are included. This suggests faculty AI policies should account for ChatGPT first, while also recognizing that many students use several tools.

Task 04: Does AI Proficiency Relate to Learning Impact?
</h3>

<p>
<strong>Question:</strong>
How does the average <code>AI_Learning_Impact</code>
differ across <code>AI_Proficiency</code> levels?
For each proficiency level, report the number of students
and the average learning impact rounded to two decimal places.
Identify the proficiency level with the highest average
learning impact.
</p>

</div>

```python
pd.read_sql_query('''
select AI_Proficiency, count(*) as students , round(avg(AI_Learning_Impact),2) AS Average_Learning_Impact 
from evaluation
group by AI_Proficiency
order by Average_Learning_Impact desc
''',conn) 

### Select count(*) show the number of students in each proficiency, 
### Round(Avg(AI_Learning_Impact),2) shows the average impact for each prodfiency, having it round to 2 decimal places
### Group by has the data organized into three groups by proficiency 
### Order by put the groups into descending order by Average learning impact
### Intermediate and Beginner AI_Proficiency levels are noted ranking the tools as the highest learning impact
```

### Business Interpretation
Intermediate (9 students) and Beginner (5 students) users tied for the highest average learning impact at **3.50**. Advanced users (6 students) reported a slightly lower average of **3.33**. The differences are small, and the groups are too small to draw strong conclusions. Still, the results suggest that greater proficiency does not necessarily mean students feel AI improves their learning more. Beginners appear to benefit just as much as more experienced users.

Task 05: Which Frequency Groups Spend More Time Using AI?
</h3>

<p>
<strong>Question:</strong>
Which <code>AI_Usage_Frequency</code> groups average
more than one hour per week using AI?
Display each qualifying group and its average
<code>AI_Hours_Per_Week</code>.
Include only groups whose average exceeds one hour.
</p>

</div>

```python
pd.read_sql_query('''
select AI_Usage_Frequency,round(avg(AI_Hours_Per_Week),2) as Average_Hours
from evaluation
group by AI_Usage_Frequency
having Average_Hours > 1
''',conn)

# group by creates one group per usage frequency level
# round(avg(AI_Hours_Per_Week), 2) gives each group's average weekly hours
# having filters the groups after averaging and keeps only those over 1 hour
# (where can't be used here because it filters rows before grouping)
```

```python
pd.read_sql_query('''
select AI_Usage_Frequency, round(avg(AI_Hours_Per_Week), 2) as Average_Hours
from evaluation
where AI_Hours_Per_Week < 10
group by AI_Usage_Frequency
having Average_Hours > 1
''', conn)

# where removes the 10-hour outlier before the groups are formed
# group by and avg() then calculate each group's average without it
# having keeps only groups whose average is still over 1 hour
```

### Business Interpretation
Two frequency groups average more than one hour per week. Students who use AI “Often” average about **3.33 hours**, and “Sometimes” users average about **1.30 hours**. The results are consistent, since students who report using AI more often also spend more time with it. However, the “Often” average may be inflated by the single 10-hour response (confirmed below and in the second query). The missing hours value is also left out of the “Sometimes” group’s average.

With the 10-hour response removed, the ‘Often’ group averages **2.0** hours per week. The outlier inflates the average, but the group still exceeds one hour and is greater than the "Sometimes" group

- **Better decisions:** Without these checks, decision-makers risk drawing conclusions from incomplete or misread data, which could lead to unfair or inaccurate admissions decisions.
