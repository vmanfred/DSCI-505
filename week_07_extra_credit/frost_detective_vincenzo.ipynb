Reads `frost.txt`, counts how often each letter a–z appears, and lists the letters from most to least frequent.

**Note:** `frost.txt` needs to be in the same folder as this notebook.

## Task 1: Read the File
`with open(...)` closes the file automatically when the block ends, so there's no need to call `.close()` separately.

```python
# Open frost.txt in read mode ("r"); the with-block closes the file automatically when it ends
with open("frost.txt", "r") as file:
    poem = file.read()  # Read the whole file into one string and store it in poem

print(poem)  # Display the poem
```

## Task 2: Prepare the Alphabet

```python
alphabet = list("abcdefghijklmnopqrstuvwxyz")  # Split the string into a list of 26 single letters
outcome = {}  # Empty dictionary that will hold letter: count pairs

print(alphabet)  # Confirm the alphabet list looks right
```

## Task 3: Letter-Counting Function
- `poem.lower()` converts the whole poem, so `S` and `s` both count as `s`.
- Any letters not in `alphabet` (spaces, digits, punctuation) fails the `if` check and is skipped.
- The function starts with an empty dictionary on every call.

```python
def count_letters(poem):
    counts = {}  # Start fresh on every call so re-running the cell doesn't double the counts

    for letters in poem.lower():  # Lowercase the poem once, then look at it one lettersacter at a time
        if letters in alphabet:  # Keep only a-z; spaces, numbers, and punctuation are skipped
            if letters not in counts:  # First time seeing this letter
                counts[letters] = 1  # Start its count at 1
            else:  # Letter has been seen before
                counts[letters] += 1  # Add 1 to its existing count (same as counts[letters] = counts[letters] + 1)

    return counts  # Return the dictionary of letter counts


outcome = count_letters(poem)  # Run the function on the poem and store the results in outcome
```

## Sort and Display
`.items()` turns the dictionary into `(letter, count)` pairs. `key=lambda item: item[1]` tells `sorted()` to sort by the count (position 1), and `reverse=True` puts the highest counts first.

```python
# Sort the (letter, count) pairs by count, highest first
sorted_counts = sorted(outcome.items(), key=lambda item: item[1], reverse=True)

print("Letter Frequencies (Most to Least):")  # Header for the results
for letter, count in sorted_counts:  # Unpack each pair into letter and count
    print(f"{letter}: {count}")  # Print in the "e 25" format the assignment shows
```

## 🏆 Bonus: Most Frequent Letter and Total Letters
Because the list is already sorted, the most frequent letter is the first item, `[0]`. Adding every count with `sum()` gives the total number of letters. The `if` check keeps an empty file from raising an error.

As a small extra, I also compared the counts against the `alphabet` list to show which letters never appear. Since the function only adds letters it actually finds, missing letters don't show up in the results above.

```python
total_letters = sum(outcome.values())  # Add up every letter's count

if sorted_counts:  # Only continue if at least one letter was found (avoids an error on an empty file)
    top_letter, top_count = sorted_counts[0]  # First pair in the sorted list is the most frequent letter
    print(f"Most frequent letter: '{top_letter}' ({top_count} times)")  # f-string inserts the values
    print(f"Total letters: {total_letters}")  # Show the total letter count
else:
    print("No letters were found.")  # Message for an empty or letter-free file

# Extra: list any letters from the alphabet that never appeared in the poem
missing = [] #empty list
for letter in alphabet: #go through letters
    if letter not in outcome: #if not in outcome
        missing.append(letter)  # Keep letters with no entry in outcome
print(f"Letters not used: {', '.join(missing) if missing else 'none'}")  # Join them into one readable line
```
